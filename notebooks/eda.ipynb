# Exploratory Data Analysis: T2D Patient Journey Cohort
### Life Sciences Commercial Analytics Practice

This notebook inspects the synthetic cohort of **N = 60** Type 2 Diabetes interview transcripts, examining baseline demographics, transcript length distributions, completeness scores, and initial exploratory cross-tabs.

```python
import json
from pathlib import Path
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

# Load extracted dataset
df = pd.read_csv("../outputs/patients_flat.csv")
print(f"Loaded {len(df)} patient records.")
df.head()
```

## 1. Demographic & Clinical Baseline Distributions

```python
print("--- Age Group Distribution ---")
print(df['age_group'].value_counts(normalize=True).round(3) * 100)

print("\n--- Disease Duration Distribution ---")
print(df['disease_duration'].value_counts(normalize=True).round(3) * 100)
```

## 2. Transcript Quality and Completeness Audit

```python
print("--- Completeness Score Summary ---")
print(df['completeness_score'].describe())

print("\n--- Partial Interview Prevalence ---")
print(df['is_partial_record'].value_counts())
```

## 3. High-Level Adoption Breakdown

```python
adoption_cross = pd.crosstab(df['disease_duration'], df['adoption_status'], margins=True)
adoption_cross
```
